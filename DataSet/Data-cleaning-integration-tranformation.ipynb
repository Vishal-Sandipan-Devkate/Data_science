# Data Transformation
is process of changing the format, structure, or value of raw data to make clean, consistent, and ready for analysis or ml modles

```python
# clean hidden leading/ trailing spaces from column names
df.columns = df.columns.str.strip()
```

```python
#Automatically find the Age and Cholesterol columns regardless of casing or exact naming
age_col = next((col for col in df.columns if "age" in col.lower()), None)
chol_col = next((col for col in df.columns if "chol" in col.lower()),None)
```

```python
print(f"Age column: {age_col}")
print(f"Cholesterol column: {chol_col}")
```

```python
# transformation age binning
if age_col:
  age_labels = ["Young", "Middle-aged", "Senior"]
  df['age_group'] = pd.cut(df[age_col], bins=[0, 30, 60, 100], labels=age_labels)
  # one-hot Encode the newly created age group
  df = pd.get_dummies(df, columns=['age_group'], dtype=int)
else:
  print("Warning: no age-related column found in dataset.")
```

```python
if chol_col:
  df["high_cholestrol"] = (df[chol_col] > 200).astype(int)
else:
  print("Warning: No cholsterol-related column found in dataset.")
```
