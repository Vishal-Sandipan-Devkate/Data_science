```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
df=pd.read_csv(r'dataset.csv')
df
```

```python
df.isna().sum()
```

```python
plt.figure(figsize=(10,10))
sns.barplot(data=df, x='Type', y='Total Interactions', palette='Blues_d', errorbar=None)
plt.title("Average Total interaction by post type:", fontsize=20, fontweight='bold')
plt.xlabel('Post Type', fontsize=12)
plt.ylabel('Average Total Interactions', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# group data by post month to see monthly data trends
monthly_trend = df.groupby('Post Month')[['Lifetime Post Total Reach','Lifetime Engaged Users', 'Total Interactions']].mean().reset_index()
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly_trend['Post Month'], monthly_trend['Lifetime Post Total Reach'], marker='o', label='Avg Lifetime Post Total Reach', color='blue')
plt.plot(monthly_trend['Post Month'], monthly_trend['Lifetime Engaged Users'], marker='s', label='Avg Lifetime Engaged Users', color='green')
plt.plot(monthly_trend['Post Month'], monthly_trend['Total Interactions'], marker='^', label='Avg Total Interactions', color='red')
plt.title('Average Facebbok performance by  month', fontsize=14, fontweight='bold')
plt.xlabel('Post Month', fontsize=12)
plt.ylabel('Average Post count', fontsize=12)
plt.legend()
plt.xticks(range(1,13))
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Total Interactions', bins=30, kde='true', color='skyblue', edgecolor='black')
plt.title('Dstribution of interaction per post', fontsize=14, fontweight='bold')
plt.xlabel('Total Interaction', fontsize=12)
plt.ylabel('Frequency post', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# Heat map
columns_correlate = ['Page total likes', 'Lifetime Post Total Reach', 'Lifetime Engaged Users', 'comment', 'like', 'share', 'Total Interactions']
corr_matrix=df[columns_correlate].corr()
```

```python
#contour plot
pivot_df=df.pivot_table(index='Post Hour', columns='Page total likes', values='Total Interactions', aggfunc='mean').fillna(0)
```

```python
# 2D coordinates
X=pivot_df.columns
Y=pivot_df.index
Z=pivot_df.values
```

```python
# plot Filled Coutour plot
plt.figure(figsize=(10,5))
coutour=plt.contourf(X, Y, Z, levels=10, cmap='viridis')
plt.colorbar(coutour, label='Average Total Interactions')
plt.title('Contour Plot of Average Total Interactions', fontsize=14, fontweight='bold')
plt.xlabel('Page total likes', fontsize=12)
plt.ylabel('Post Hour', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# Stem plot
data_sample = df.head(30)
x=range(len(data_sample))
y=data_sample['Lifetime Post Total Reach']

```

```python
plt.figure(figsize=(10,5))
plt.stem(x, y, linefmt='blue', markerfmt='ro', basefmt='gray')
plt.title('Stem plot of Lifetime Post Total Reach(30 samples)', fontsize=14, fontweight='bold')
plt.xlabel('Post Index')
plt.ylabel('Total Reach')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import numpy as np
flow_data = (df.groupby(['Post Month', 'Post Hour'])[['Lifetime Post Total Reach', 'Total Interactions']].mean().reset_index())
X = flow_data['Post Hour']
Y = flow_data['Post Month']
U = flow_data['Lifetime Post Total Reach'] / 1000
V = flow_data['Total Interactions'] / 100
magnitude = np.hypot(U, V)
```

```python
plt.figure(figsize=(10, 6))
quiver = plt.quiver(X, Y, U, V,magnitude, cmap='viridis', scale=50, pivot='mid')
plt.colorbar(quiver, label='Engagement Flow Magnitude')
plt.title('Engagement Flow Plot (Reach & Interactions by Hour and Month)',fontsize=14, fontweight='bold')
plt.xlabel('Post Hour (0 to 23)', fontsize=12)
plt.ylabel('Post Month (1 to 12)', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
hourly_data = df.groupby('Post Hour')['Total Interactions'].mean()
hourly_data = hourly_data.reindex(range(24), fill_value=0)
theta = np.linspace(0, 2 * np.pi, 24, endpoint=False)
r = hourly_data.values
```

```python
plt.figure(figsize=(8, 8))
ax = plt.subplot(111, projection='polar')
ax.plot(theta, r, color='crimson', linewidth=2, label='Avg Interactions')
ax.fill(theta, r, color='crimson', alpha=0.25)
ax.set_theta_zero_location('N')
ax.set_theta_direction(-1)
ax.set_xticks(theta)
ax.set_xticklabels([f'{h}:00' for h in range(24)], fontsize=8)
plt.title('Facebook Engagement Across 24 Hours (Polar View)', fontsize=14, fontweight='bold', pad=20)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```
